```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv") # df is data frame and i am importing the iris data set
```

```python
df.head() # not using print(df) because it prints full data that to without table
```

```python
df.columns = ['SL', 'SW', 'PL', 'PW', 'species']
```

```python
plt.scatter(df['SL'], df['SW']) #plot a scatterplot
```

```python
df.species.unique() # listing the different values of species
```

```python
colors = {'setosa':'blue','versicolor':'red','virginica':'green'} # {} creates a dictionary - mapping features to values
```

```python
plt.scatter(df['SL'], df['SW'], c=df['species'].map(colors)) # c is a parameter of plt for color
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=['species']) # copy table without species column
```

```python
y = df['species'] # just the species column
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.7)
```

```python
x_train.head() # looking at head
```

```python
y_train.head()
```

```python
import seaborn as sns
```

```python
sns.pairplot(df, hue='species') #lot of charts - fuller data visualization
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf = DecisionTreeClassifier()
```

```python
clf.fit(x_train, y_train) # fitting a df in a classifier (model) - the model has been trained now
```

```python
y_pred = clf.predict(x_test) # on the graph we are about to plot, x is on Yaxis and y is on Xaxis
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```

```python
# Data frames used:
# X_train: Training features, model learns from this.
# y_train: Training labels, correct answers for training.
# X_test: Testing features, used to make predictions.
# y_test: Testing labels, correct answers to measure accuracy.
# y_pred: The model's predicted labels for X_test
```
